```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
```

## Get the Data 

```python
#https://www.kaggle.com/uciml/iris

url_='https://drive.google.com/uc?id=10vsX925g8ff4QdgQpnsguo4DHUHOPgAL'
df = pd.read_csv(url_)
```

```python
X = df.drop(['Id', 'Species'], axis = 1)
y = df['Species']
```

## Split data to Train and Test Sets

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20) # 20%
```

```python
(X_train.shape, y_train.shape)
```

```python
(X_test.shape, y_test.shape)
```

### Decision Tree - create instance

```python
# https://scikit-learn.org/stable/modules/generated/sklearn.tree.DecisionTreeClassifier.html
model = DecisionTreeClassifier(criterion='gini')
```

### Decision Tree - train model

```python
model.fit(X_train, y_train)
```

### Decision Tree - make a prediction

```python
y_pred = model.predict(X_test)
y_pred_s = pd.Series(y_pred)
```

```python
type(y_pred)
```

```python
y_pred_s
```

```python
type(y_test)
```

```python
# combine to view
s3 = pd.Series(list(zip(y_pred_s, y_test)))
s3
```

## Model Evaluation

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(confusion_matrix(y_test, y_pred)) 
```

## Visualization

```python
plt.figure(figsize = (10, 10))
plot_tree(model, feature_names = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm'], 
          class_names = ['Iris-setosa', 'Iris-versicolor', 'Iris-virginica'], filled = True)
plt.show()
```

## Продивився даний ноутбук. 


Ну шо, у підсумку можна сказати, що у цьому ноутбуку завантажується датасет Iris, обираються чотири числові ознаки (довжина й ширина чашолистка та пелюстки), а ціль - вид ірису з трьох класів. Дані діляться на тренувальну і тестову вибірки у співвідношенні 80/20, після чого тренуються і порівнюються три класичні алгоритми класифікації з бібліотеки scikit‑learn: дерево рішень, k‑найближчих сусідів та опорні вектори. 

Для дерева рішень створюється екземпляр із критерієм Джині, модель навчається на X_train, y_train, робляться передбачення на X_test, а якість оцінюється за звітом класифікації і матрицею змішування: точність по класах, повнота, f1 та загальна точність ~0.9 на випадковому розбитті. 

Додатково візуалізується структура дерева через plot_tree з підписами ознак і класів. Аналогічно для kNN підбирається k і порівнюються метрики на тесті, де kNN добре працює завдяки чіткій кластеризації класів у Iris. Для SVM запускаються як лінійне ядро (plain), так і радіально‑базисне (RBF): лінійне показує високу якість, коли межі між класами майже лінійно відокремні, а RBF захоплює слабкі нелінійності між versicolor і virginica; обидва варіанти оцінюються тими ж метриками. Підсумково ноутбук демонструє повний цикл: завантаження даних, формування X/y, розбиття, навчання трьох моделей. 

На останок порівняння за класифікаційним звітом і матрицею змішування, + інтерпретована візуалізацію для дерева.

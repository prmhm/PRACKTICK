```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# generation of synthetic dataset
N = 100
a = 1
b = 3

x = np.random.uniform(0, 5, N)
noise = np.random.normal(0, .5, N)
y = a * x + b + noise

plt.grid()
plt.scatter(x, y, marker = '.', color = 'r')
```

```python
X = x.reshape(-1, 1)
```

```python
x # 1d array
```

```python
X # array where each row is an array to
```

```python
# a * x + b = y

model = LinearRegression()
model.fit(X, y)
model
```

```python
a_found = model.coef_[0]
b_found = model.intercept_
```

Класична лінійна регресія з бібліотеки sklearn; включає створення синтетичного датасету з шумом, навчання лінійної моделі, отримання коефіцієнтів, візуалізація результатів з порівнянням оригінальної функції та знайденої моделі

```python
#Додам лінійну функцію з шумом
x = np.linspace(0, 10, 50)
y_ = 2 * x + 3  #Чиста лінійна функція
noise = np.random.normal(0, 2, len(x))#шум
y_noisy = y_ + noise  #Дані з шумом
```

Візуалізація

```python
plt.figure(figsize=(10, 6))
#Графік лін. функції
plt.plot(x, y_, '--', color='blue', linewidth=2, label='Чиста лінійна функція (y = 2x + 3)')
#Точки з шумом
plt.scatter(x, y_noisy, color='red', alpha=0.6, s=50, label='Дані з шумом')
```

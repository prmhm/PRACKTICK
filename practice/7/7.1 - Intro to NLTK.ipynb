## Stop words filtration

```python
stop_words = stopwords.words('english')
stop_words
```

```python
text = "This is a sample sentence, showing off the stop words filtration."
text
```

```python
from nltk import FreqDist
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import matplotlib.pyplot as plt

def plot_top_words(text, n=20):
    stop = set(stopwords.words("english"))
    tokens = [w.lower() for w in word_tokenize(text) if w.isalpha()]
    tokens = [w for w in tokens if w not in stop]
    fdist = FreqDist(tokens)
    words, counts = zip(*fdist.most_common(n))
    plt.figure(figsize=(8,4))
    plt.bar(words, counts)
    plt.xticks(rotation=60, ha="right")
    plt.ylabel("Frequency")
    plt.title(f"Top {n} content words")
    plt.tight_layout()
    plt.show()
```

```python
from nltk.book import text1
plot_top_words(" ".join(text1[:5000]), n=20)
#додав візуалізацію частот найуживаніших змістовних слів у тексті 1:
```

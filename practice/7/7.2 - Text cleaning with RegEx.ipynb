```python
import re
import string
```

```python
text = 'Noisy Text Example with a lot of numbers and special symbols: 12345, #abc, @@qwerty, http://xyz. Data cleaning Done'
```

```python
# Converting to lowercase
text = text.lower()
text
```

```python
# remove mentions
text = re.sub("@\\S+", " ", text)
text
```

```python
# remove url
text = re.sub("https*\\S+", " ", text)
text
```

```python
# remove hashtags
text = re.sub("#\\S+", " ", text)
text
```

```python
# remove all numbers
text = re.sub("\\d", " ", text)  
text
```

```python
# remove punctuations
text = re.sub('[%s]' % re.escape(string.punctuation), ' ', text)
text
```

```python
# remove extra spaces
text = re.sub('\\s{2,}', " ", text)
text
```

```python
from nltk.corpus import stopwords
stop_words = stopwords.words("english")
text = ' '.join([word for word in text.split(' ') if word not in stop_words])
text
```

```python
# example of text cleaning function
def clean_text(text):
    text = text.lower()
    text = re.sub("@\\S+", " ", text)
    text = re.sub("https*\\S+", " ", text)
    text = re.sub("#\\S+", " ", text)
    text = re.sub("\\d", " ", text)
    text = re.sub('[%s]' % re.escape(string.punctuation), ' ', text)
    text = re.sub('\\n', ' ', text)
    text = re.sub('\\s{2,}',' ', text)
    text = ' '.join([word for word in text.split(' ') if word not in stop_words])
    return text
```

Тут пайплайн очищення тексту. Він за допомогою регулярних виразів і стоп-слів, на наприкінці загортає всі кроки в універсальну функцію clean_text.

Спочатку сирий рядок приводиться до нижнього регістру, щоб позбутися чутливості до регістру, далі послідовно видаляються згадки користувачів @username, URL‑адреси, хештеги всякі (#текст), усі цифри, пунктуація зі string.punctuation, переводи рядка та «злиплі» багаторазові пробіли — усе це робиться через метод re.sub.
Тобто  у підсумку функція clean_text приймає довільний рядок, проганяє його через усі ці кроки і повертає максимально чистий текст, готовий до токенізації, побудови ознак або навчання моделей NLP.

Тепер зберу все це в один блок та додам статистику: 

```python
def clean_text_with_stats(text):
    original_length = len(text)
    original_words = len(text.split())
    
    # Підрахунок видалених елементів
    mentions = len(re.findall("@\\S+", text))
    urls = len(re.findall("https*\\S+", text))
    hashtags = len(re.findall("#\\S+", text))
    numbers = len(re.findall("\\d+", text))
    
    # Очищення
    cleaned = clean_text(text)
    
    # Статистика
    stats = {
        'original_length': original_length,
        'cleaned_length': len(cleaned),
        'original_words': original_words,
        'cleaned_words': len(cleaned.split()),
        'removed_mentions': mentions,
        'removed_urls': urls,
        'removed_hashtags': hashtags,
        'removed_numbers': numbers,
        'compression_ratio': f"{(1 - len(cleaned)/original_length)*100:.1f}%"
    }
    
    return cleaned, stats

#використання:
text = 'Check this out! #AI #MachineLearning @john_doe https://example.com 12345 @@@ amazing!!!'
cleaned, stats = clean_text_with_stats(text)

print("Оригінал:", text)
print("\nОчищений:", cleaned)
print("\nСтатистика:")
for key, value in stats.items():
    print(f"  {key}: {value}")
```
